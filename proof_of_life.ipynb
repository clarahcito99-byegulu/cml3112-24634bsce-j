My explanation: The cell `readings = readings + 5` does not set readings to a fixed value; it takes whatever value readings currently holds in the kernel's memory and adds 5 to it. Running it once gives 17 (12+5), running it again gives 22 (17+5), and a third time gives 27 (22+5), even though the code itself never changed. This happens because all cells in a notebook share one kernel, so the kernel keeps the last value of readings between runs. After Restart and Run All, the kernel is cleared and every cell runs once from the top in order, so readings goes back to starting at 12 and the cell only adds 5 once, giving 17. This is why Restart and Run All is the only honest test of a notebook: it proves the result comes from the code as written, not from leftover values from cells run out of order.
