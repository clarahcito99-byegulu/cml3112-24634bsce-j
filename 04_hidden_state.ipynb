**Predicted result before repair:** prints 12, not 17, even though the comment says 17 is expected.

**Cause:** there was a hidden extra line, `readings = 12`, placed after the `+ 5` step and before the `print`. It silently resets `readings` back to 12 right before printing, overwriting the addition. No error is raised, so the bug is easy to miss on a quick glance.

**Repair:** removed the extra `readings = 12` reset line, so the sequence is just start at 12, add 5, then print - giving 17 as expected.
